```python
import pandas as pd

# Rutas relativas a la raíz del proyecto (funcionan en cualquier equipo)
ruta_entrada = "../data/raw/resultados_analisis_sentimientos_hoteles.csv"
ruta_salida = "../data/muestra/resenas_muestra.csv"

df = pd.read_csv(ruta_entrada)
df.columns = df.columns.str.strip()

# Se eliminan las columnas que identifican a los autores de las reseñas
df = df.drop(columns=['autor', 'URL del autor', 'URL de revisión', 'ID de revisión'])

df.sample(200, random_state=42).to_csv(ruta_salida, index=False)

muestra = df.sample(200, random_state=42)
muestra.to_csv(ruta_salida, index=False)

print(f"Muestra generada: {len(muestra)} filas")
print(f"Columnas conservadas: {list(muestra.columns)}")
```
